```python
from pathlib import Path
import xarray as xr
import pandas as pd

var_dir = Path("data/argo_monthly_nc")
files = sorted(var_dir.glob("argo_bgc_global_*.nc"))

profile_rows = []

for f in files:
    print(f"Reading {f}")
    ds = xr.open_dataset(f)

    # Select only the columns we need for the profile index
    needed = ["PLATFORM_NUMBER", "CYCLE_NUMBER", "TIME", "LATITUDE", "LONGITUDE", "CHLA_QC", "PRES"]
    missing = [v for v in needed if v not in ds.variables and v not in ds.coords]
    if missing:
        raise ValueError(f"{f} missing required vars: {missing}")

    df = (
        ds[needed]
        .to_dataframe()
        .reset_index(drop=True)
    )

    # Drop any rows without platform/cycle info
    df = df.dropna(subset=["PLATFORM_NUMBER", "CYCLE_NUMBER"])

    # Collapse to ONE row per (PLATFORM_NUMBER, CYCLE_NUMBER).
    # We just keep:
    #   - TIME: first
    #   - LAT/LON: first (typically constant within profile)
    df_profiles = (
        df
        .sort_values("TIME")  # just to make "first" deterministic
        .groupby(["PLATFORM_NUMBER", "CYCLE_NUMBER"], as_index=False)
        .agg({
            "TIME": "first",
            "LATITUDE": "first",
            "LONGITUDE": "first",
        })
    )

    profile_rows.append(df_profiles)

# Combine all months
profiles = pd.concat(profile_rows, ignore_index=True)

# If there is overlap between months, enforce uniqueness
profiles = (
    profiles
    .sort_values("TIME")
    .drop_duplicates(subset=["PLATFORM_NUMBER", "CYCLE_NUMBER"], keep="first")
    .reset_index(drop=True)
)

# Create a stable profile_id: e.g. "6901234_0042"
profiles["profile_id"] = (
    profiles["PLATFORM_NUMBER"].astype(int).astype(str).str.zfill(7)
    + "_"
    + profiles["CYCLE_NUMBER"].astype(int).astype(str).str.zfill(4)
)

# Reorder columns: profile_id first, then metadata
profiles = profiles[
    ["profile_id", "PLATFORM_NUMBER", "CYCLE_NUMBER", "TIME",
     "LATITUDE", "LONGITUDE"]
]

print(len(profiles), "unique profiles")
print(profiles.columns)

# Save for later joins
out_path = var_dir / "argo_profiles_index.parquet"
profiles.to_parquet(out_path, index=False)
print("Wrote", out_path)
```

```python
from pathlib import Path
import numpy as np
import pandas as pd

var_dir = Path("data/matchups")
var = "chlor_a"
var = "Kd"
var = "Rrs"

files = sorted(var_dir.glob(f"chla_profile_matchups_{var}_batch_*.parquet"))

def pick_best_L3_per_profile(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()

    # --- 1. Make sure the PACE times are proper UTC datetimes ---
    df[f"pace_{var}_t_start"] = pd.to_datetime(df[f"pace_{var}_t_start"], utc=True)
    df[f"pace_{var}_t_end"]   = pd.to_datetime(df[f"pace_{var}_t_end"],   utc=True)

    # Argo TIME is naive datetime64[ns] in your example; assume it's UTC and localize
    if pd.api.types.is_datetime64_ns_dtype(df["TIME"]):
        df["TIME_utc"] = df["TIME"].dt.tz_localize("UTC")
    else:
        df["TIME_utc"] = pd.to_datetime(df["TIME"], utc=True)

    # --- 2. Compute midpoint time of each L3 granule & time-diff to Argo ---
    df["pace_L3_time"] = df[f"pace_{var}_t_start"] + (
        df[f"pace_{var}_t_end"] - df[f"pace_{var}_t_start"]
    ) / 2

    df["time_diff"] = (df["pace_L3_time"] - df["TIME_utc"]).abs()

    # --- 3. For each profile, keep the L3 row with smallest time_diff ---
    group_cols = ["PLATFORM_NUMBER", "CYCLE_NUMBER", "TIME"]
    df_best = (
        df.sort_values("time_diff")
          .groupby(group_cols, as_index=False)
          .first()
    )

    # Drop helper columns we don't need in the final product
    df_best = df_best.drop(columns=["TIME_utc", "time_diff"])

    # --- 4. Rename pace_chlor_* columns to pace_L3_*, with special case for the variable ---
    # All pace-related columns except the data variable itself
    rename_dict = {
        col: col.replace(f"pace_{var}", "pace_L3")
        for col in df_best.columns
        if col.startswith(f"pace_{var}") and col != f"pace_{var}" and not col[-1].isdigit()
    }
    df_best = df_best.rename(columns=rename_dict)    
    # Now rename the variable itself
    rename_dict = {
        col: col.replace(f"pace_{var}", f"pace_L3_{var}")
        for col in df_best.columns
        if col.startswith(f"pace_{var}")
    }
    df_best = df_best.rename(columns=rename_dict)    

    if "pace_L3_file" in df_best.columns:
        df_best = df_best.rename(columns={"pace_L3_file": f"pace_L3_{var}_file"})

    # At this point you should have:
    #   - Argo metadata unchanged (PLATFORM_NUMBER, CYCLE_NUMBER, TIME, LATITUDE, LONGITUDE, CHLA_*…)
    #   - pace_L3_file, pace_L3_t_start, pace_L3_t_end, pace_L3_lat, pace_L3_lon
    #   - pace_L3_time (midpoint)
    #   - pace_L3_chlor_a (the variable you care about)
    # Create a stable profile_id: e.g. "6901234_0042"
    df_best["profile_id"] = (
        profiles["PLATFORM_NUMBER"].astype(int).astype(str).str.zfill(7)
        + "_"
        + profiles["CYCLE_NUMBER"].astype(int).astype(str).str.zfill(4)
    )
    
    # Reorder columns: profile_id first, then metadata
    prefix = f"pace_L3_{var}"
    pace_var_cols = [c for c in df_best.columns if c.startswith(prefix)]
    cols = ["profile_id", "PLATFORM_NUMBER", "CYCLE_NUMBER", "TIME", "LATITUDE", "LONGITUDE",
           'pace_L3_time', 'pace_L3_t_start', 'pace_L3_t_end','pace_L3_lat', 'pace_L3_lon']
    cols.extend(pace_var_cols)
    df_best = df_best[ cols ]

    return df_best


# If you want to aggregate all batches into one file:
dfs = []
for f in files:
    df_batch = pd.read_parquet(f)
    dfs.append(pick_best_L3_per_profile(df_batch))

df_all_best = pd.concat(dfs, ignore_index=True)
df_all_best = df_all_best.sort_values("TIME")
df_all_best.to_parquet(f"data/matchups/pace_L3_{var}_matchups.parquet", index=False)
```
